```python
import numpy as np
import pandas as pd
import plotly.io as pio
from scipy.spatial import cKDTree


def idw_grid(
    df,
    x_col="X",
    y_col="Y",
    value_col="VALUE",
    nx=150,
    ny=150,
    power=2.0,
    k=8,
    max_distance=None,
    padding=0.0,
):
    """
    Interpolate scattered points onto a regular grid using IDW.

    Parameters
    ----------
    df : pandas.DataFrame
        Must contain x_col, y_col, value_col.
    nx, ny : int
        Number of grid nodes in X and Y.
    power : float
        IDW power. Higher values make nearby points dominate more.
    k : int
        Number of nearest points used for each grid node.
    max_distance : float or None
        If given, grid nodes farther than this from the nearest point are set to NaN.
    padding : float
        Extra margin around min/max coordinates.
    """

    x = df[x_col].to_numpy(dtype=float)
    y = df[y_col].to_numpy(dtype=float)
    z = df[value_col].to_numpy(dtype=float)

    valid = np.isfinite(x) & np.isfinite(y) & np.isfinite(z)
    x, y, z = x[valid], y[valid], z[valid]

    if len(z) == 0:
        raise ValueError("No valid x/y/value rows found.")

    xi = np.linspace(x.min() - padding, x.max() + padding, nx)
    yi = np.linspace(y.min() - padding, y.max() + padding, ny)
    Xg, Yg = np.meshgrid(xi, yi)

    grid_points = np.column_stack([Xg.ravel(), Yg.ravel()])
    data_points = np.column_stack([x, y])

    tree = cKDTree(data_points)

    k = min(k, len(z))
    distances, indexes = tree.query(grid_points, k=k)

    if k == 1:
        distances = distances[:, None]
        indexes = indexes[:, None]

    nearest_distance = distances[:, 0]

    # Exact observation hit: avoid division by zero
    eps = 1e-12
    weights = 1.0 / np.maximum(distances, eps) ** power

    values = z[indexes]
    zg = np.sum(weights * values, axis=1) / np.sum(weights, axis=1)

    if max_distance is not None:
        zg[nearest_distance > max_distance] = np.nan

    Zg = zg.reshape(Xg.shape)

    return Xg, Yg, Zg


# ------------------------------------------------
# Example test data
# ------------------------------------------------
np.random.seed(42)

df = pd.DataFrame({
    "X": np.random.uniform(-500, 500, 50),
    "Y": np.random.uniform(-1200, 1200, 50),
    "VALUE": np.random.uniform(0, 100, 50),
})

Xg, Yg, Zg = idw_grid(
    df,
    nx=180,
    ny=180,
    power=4.0,
    k=5,
    max_distance=550,
)

fig = {
    "data": [
        {
            "type": "heatmap",
            "x": Xg[0, :].tolist(),
            "y": Yg[:, 0].tolist(),
            "z": Zg.tolist(),
            "colorscale": "Viridis",
            "name": "IDW grid",
            "colorbar": {"title": "Value"},
        },
        {
            "type": "scatter",
            "x": df["X"].tolist(),
            "y": df["Y"].tolist(),
            "mode": "markers",
            "name": "Original points",
            "marker": {
                "size": 8,
                "color": "black",
                "line": {"width": 1, "color": "white"},
            },
        },
    ],
    "layout": {
        "title": "IDW Interpolation with Distance Mask",
        "xaxis": {"title": "X", "scaleanchor": "y"},
        "yaxis": {"title": "Y"},
        "template": "plotly_white",
        "width": 900,
        "height": 700,
    },
}

pio.show(fig)
```

```python
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree
import plotly.io as pio
from scipy.spatial import Delaunay

def apply_convex_hull_mask(Xg, Yg, df, grids, x_col="X", y_col="Y"):
    points = df[[x_col, y_col]].to_numpy(dtype=float)
    grid_points = np.column_stack([Xg.ravel(), Yg.ravel()])

    hull = Delaunay(points)
    inside = hull.find_simplex(grid_points) >= 0
    inside = inside.reshape(Xg.shape)

    masked_grids = []
    for grid in grids:
        g = grid.copy()
        g[~inside] = np.nan
        masked_grids.append(g)

    return masked_grids, inside

def idw_grid(
    df,
    x_col="X",
    y_col="Y",
    value_col="VALUE",
    nx=180,
    ny=180,
    power=2.0,
    k=8,
    max_distance=None,
):
    x = df[x_col].to_numpy(dtype=float)
    y = df[y_col].to_numpy(dtype=float)
    z = df[value_col].to_numpy(dtype=float)

    valid = np.isfinite(x) & np.isfinite(y) & np.isfinite(z)
    x, y, z = x[valid], y[valid], z[valid]

    xi = np.linspace(x.min(), x.max(), nx)
    yi = np.linspace(y.min(), y.max(), ny)
    Xg, Yg = np.meshgrid(xi, yi)

    grid_points = np.column_stack([Xg.ravel(), Yg.ravel()])
    data_points = np.column_stack([x, y])

    tree = cKDTree(data_points)

    k = min(k, len(z))
    distances, indexes = tree.query(grid_points, k=k)

    if k == 1:
        distances = distances[:, None]
        indexes = indexes[:, None]

    eps = 1e-12
    weights = 1.0 / np.maximum(distances, eps) ** power

    values = z[indexes]
    zg = np.sum(weights * values, axis=1) / np.sum(weights, axis=1)

    if max_distance is not None:
        nearest_distance = distances[:, 0]
        zg[nearest_distance > max_distance] = np.nan

    return Xg, Yg, zg.reshape(Xg.shape)


def phase_rgb_from_dominance(oil_g, water_g, gas_g):
    """
    Option 3:
    - Hue/colour = dominant phase
    - Brightness/purity = dominance strength
    """

    stack = np.stack([oil_g, water_g, gas_g], axis=0)

    dominant = np.nanargmax(stack, axis=0)
    purity = np.nanmax(stack, axis=0)

    # Vertex colours
    oil_color = np.array([160, 160, 160], dtype=float)   # grey
    water_color = np.array([0, 90, 255], dtype=float)    # blue
    gas_color = np.array([150, 90, 40], dtype=float)     # brown

    colors = np.stack([oil_color, water_color, gas_color], axis=0)

    rgb = colors[dominant]

    # Make mixed regions paler.
    # purity=1.0 -> full colour
    # purity=1/3 -> almost white
    purity_scaled = (purity - 1/3) / (1 - 1/3)
    purity_scaled = np.clip(purity_scaled, 0, 1)

    white = np.array([255, 255, 255], dtype=float)

    rgb = white + purity_scaled[..., None] * (rgb - white)

    invalid = np.isnan(oil_g) | np.isnan(water_g) | np.isnan(gas_g)
    rgb[invalid] = [255, 255, 255]

    return np.clip(rgb, 0, 255).astype(np.uint8), dominant, purity


# ------------------------------------------------
# 1. Synthetic scattered data
# ------------------------------------------------
np.random.seed(42)

n_points = 50

df = pd.DataFrame({
    "X": np.random.uniform(-500, 500, n_points),
    "Y": np.random.uniform(-1200, 1200, n_points),
})

# Random oil/water/gas fractions that sum to 1
fractions = np.random.dirichlet(alpha=[2, 2, 2], size=n_points)

df["OIL"] = fractions[:, 0]
df["WATER"] = fractions[:, 1]
df["GAS"] = fractions[:, 2]


# ------------------------------------------------
# 2. Interpolate each fraction independently
# ------------------------------------------------
params = {
    "nx": 200,
    "ny": 200,
    "power": 2.0,
    "k": 8,
    "max_distance": 350,
}

Xg, Yg, Oil_g = idw_grid(df, value_col="OIL", **params)
_, _, Water_g = idw_grid(df, value_col="WATER", **params)
_, _, Gas_g = idw_grid(df, value_col="GAS", **params)

# Re-normalize fractions at each grid node
total = Oil_g + Water_g + Gas_g
Oil_g = Oil_g / total
Water_g = Water_g / total
Gas_g = Gas_g / total


# ------------------------------------------------
# 3. Convert dominant phase + purity to RGB
# ------------------------------------------------
RGB_g, dominant_g, purity_g = phase_rgb_from_dominance(
    Oil_g,
    Water_g,
    Gas_g
)


# ------------------------------------------------
# 4. Plotly JSON figure
# ------------------------------------------------
fig = {
    "data": [
        {
            "type": "image",
            "z": RGB_g.tolist(),
            "x0": float(Xg[0, 0]),
            "y0": float(Yg[0, 0]),
            "dx": float(Xg[0, 1] - Xg[0, 0]),
            "dy": float(Yg[1, 0] - Yg[0, 0]),
            "name": "Dominant phase + purity",
        },
        {
            "type": "scatter",
            "x": df["X"].tolist(),
            "y": df["Y"].tolist(),
            "mode": "markers",
            "name": "Original samples",
            "marker": {
                "size": 9,
                "color": "black",
                "line": {"width": 1, "color": "white"},
            },
            "text": [
                f"Oil={o:.2f}<br>Water={w:.2f}<br>Gas={g:.2f}"
                for o, w, g in zip(df["OIL"], df["WATER"], df["GAS"])
            ],
            "hovertemplate": (
                "X: %{x:.1f}<br>"
                "Y: %{y:.1f}<br>"
                "%{text}<extra></extra>"
            ),
        },
    ],
    "layout": {
        "title": "Dominant Fluid Phase with Purity",
        "xaxis": {
            "title": "X",
            "scaleanchor": "y",
        },
        "yaxis": {
            "title": "Y",
        },
        "width": 900,
        "height": 750,
        "template": "plotly_white",
        "annotations": [
            {
                "text": "Grey = Oil dominant | Blue = Water dominant | Brown = Gas dominant<br>Paler colour = more mixed composition",
                "xref": "paper",
                "yref": "paper",
                "x": 0.5,
                "y": -0.12,
                "showarrow": False,
            }
        ],
    },
}

pio.show(fig)
```

```python
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree, Delaunay
import plotly.io as pio


def idw_grid(df, value_col, x_col="X", y_col="Y", nx=220, ny=220, power=2.0, k=8):
    x = df[x_col].to_numpy(dtype=float)
    y = df[y_col].to_numpy(dtype=float)
    z = df[value_col].to_numpy(dtype=float)

    xi = np.linspace(x.min(), x.max(), nx)
    yi = np.linspace(y.min(), y.max(), ny)
    Xg, Yg = np.meshgrid(xi, yi)

    grid_points = np.column_stack([Xg.ravel(), Yg.ravel()])
    data_points = np.column_stack([x, y])

    tree = cKDTree(data_points)
    k = min(k, len(z))

    distances, indexes = tree.query(grid_points, k=k)

    if k == 1:
        distances = distances[:, None]
        indexes = indexes[:, None]

    eps = 1e-12
    weights = 1.0 / np.maximum(distances, eps) ** power

    values = z[indexes]
    zg = np.sum(weights * values, axis=1) / np.sum(weights, axis=1)

    return Xg, Yg, zg.reshape(Xg.shape)


def apply_convex_hull_mask(Xg, Yg, df, grids, x_col="X", y_col="Y"):
    points = df[[x_col, y_col]].to_numpy(dtype=float)
    grid_points = np.column_stack([Xg.ravel(), Yg.ravel()])

    hull = Delaunay(points)
    inside = hull.find_simplex(grid_points) >= 0
    inside = inside.reshape(Xg.shape)

    masked = []
    for grid in grids:
        g = grid.copy()
        g[~inside] = np.nan
        masked.append(g)

    return masked, inside


def ternary_rgb(oil, water, gas):
    """
    Triangle / barycentric colour interpolation.

    Oil   = grey
    Water = blue
    Gas   = brown

    RGB = oil * grey + water * blue + gas * brown
    """
    oil_color = np.array([160, 160, 160], dtype=float)
    water_color = np.array([0, 90, 255], dtype=float)
    gas_color = np.array([150, 90, 40], dtype=float)

    total = oil + water + gas

    oil = oil / total
    water = water / total
    gas = gas / total

    rgb = (
        oil[..., None] * oil_color +
        water[..., None] * water_color +
        gas[..., None] * gas_color
    )

    invalid = np.isnan(oil) | np.isnan(water) | np.isnan(gas)
    rgb[invalid] = [255, 255, 255]

    return np.clip(rgb, 0, 255).astype(np.uint8)


# ------------------------------------------------
# 1. Synthetic scattered data
# ------------------------------------------------
np.random.seed(42)

n_points = 50

df = pd.DataFrame({
    "X": np.random.uniform(-500, 500, n_points),
    "Y": np.random.uniform(-1200, 1200, n_points),
})

fractions = np.random.dirichlet(alpha=[2, 2, 2], size=n_points)

df["OIL"] = fractions[:, 0]
df["WATER"] = fractions[:, 1]
df["GAS"] = fractions[:, 2]


# ------------------------------------------------
# 2. Interpolate each fraction separately
# ------------------------------------------------
params = {
    "nx": 220,
    "ny": 220,
    "power": 2.0,
    "k": 8,
}

Xg, Yg, Oil_g = idw_grid(df, value_col="OIL", **params)
_, _, Water_g = idw_grid(df, value_col="WATER", **params)
_, _, Gas_g = idw_grid(df, value_col="GAS", **params)


# ------------------------------------------------
# 3. Mask outside convex hull
# ------------------------------------------------
[Oil_g, Water_g, Gas_g], inside_hull = apply_convex_hull_mask(
    Xg,
    Yg,
    df,
    [Oil_g, Water_g, Gas_g],
)


# ------------------------------------------------
# 4. Convert interpolated fractions to RGB by triangle interpolation
# ------------------------------------------------
RGB_g = ternary_rgb(Oil_g, Water_g, Gas_g)


# ------------------------------------------------
# 5. Plotly JSON figure
# ------------------------------------------------
fig = {
    "data": [
        {
            "type": "image",
            "z": RGB_g.tolist(),
            "x0": float(Xg[0, 0]),
            "y0": float(Yg[0, 0]),
            "dx": float(Xg[0, 1] - Xg[0, 0]),
            "dy": float(Yg[1, 0] - Yg[0, 0]),
            "name": "Ternary fluid colour",
        },
        {
            "type": "scatter",
            "x": df["X"].tolist(),
            "y": df["Y"].tolist(),
            "mode": "markers",
            "name": "Original samples",
            "marker": {
                "size": 9,
                "color": "black",
                "line": {"width": 1, "color": "white"},
            },
            "text": [
                f"Oil={o:.2f}<br>Water={w:.2f}<br>Gas={g:.2f}"
                for o, w, g in zip(df["OIL"], df["WATER"], df["GAS"])
            ],
            "hovertemplate": (
                "X: %{x:.1f}<br>"
                "Y: %{y:.1f}<br>"
                "%{text}<extra></extra>"
            ),
        },
    ],
    "layout": {
        "title": "Triangle / Ternary Colour Interpolation: Oil-Water-Gas",
        "xaxis": {
            "title": "X",
            "scaleanchor": "y",
        },
        "yaxis": {
            "title": "Y",
        },
        "width": 900,
        "height": 750,
        "template": "plotly_white",
        "annotations": [
            {
                "text": (
                    "Oil = grey | Water = blue | Gas = brown<br>"
                    "Each grid colour = oil·grey + water·blue + gas·brown"
                ),
                "xref": "paper",
                "yref": "paper",
                "x": 0.5,
                "y": -0.12,
                "showarrow": False,
            }
        ],
    },
}

pio.show(fig)
```
